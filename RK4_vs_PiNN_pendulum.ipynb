```python
labels = [
    "first working",
    "v1 commit",
    "v2 commit",
    "v3-1opt",
    "omp implement",
    "-ffast-math"
]

times = [5.0, 0.5, 0.017, 0.0116, 0.00488, 0.00358]

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(labels, times, marker="o", color="pink", linewidth=2.5)

ax.set_yscale("log")
ax.set_ylabel("time per iteration (s)")

for x, t in zip(labels, times):
    ax.annotate(f"{t} s", (x, t), textcoords="offset points",
                xytext=(5, 8), fontsize=9)

ax.annotate("10x", (0.5, 1.5), ha="center", color="gray")
ax.annotate("29.4x", (1.5, 0.09), ha="center", color="gray")
ax.annotate("1.47x", (2.5, 0.014), ha="center", color="gray")
ax.annotate("2.42x", (3.5, 0.007), ha="center", color="gray")
ax.annotate("1.36x", (4.5, 0.004), ha="center", color="gray")

ax.grid(True, which="both", axis="y", alpha=0.3)
plt.xticks(rotation=15, ha="right")
plt.tight_layout()
plt.show()
```

```python
print("Speedup from first working to current -> ", times[0]/times[-1],"times faster")
```
